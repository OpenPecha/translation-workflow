```python
import pandas as pd 
```

```python
data= pd.read_json("choenjuk_with_commentaries.json")
data_2= pd.read_json("choenjuk.jsonl",lines=True)
```

```python
data.rename(columns={"root_display_text": "source"}, inplace=True)
```

```python
data_2['source'] = data_2['source'].astype(str)
data['source'] = data['source'].astype(str)

# Check for common values in source columns

common_sources = set(data['source']).intersection(set(data_2['source']))
print("\nNumber of common sources:", len(common_sources))
print("Sample of common sources:", list(common_sources)[:5] if common_sources else "No common sources found")
```

```python
joined_source = pd.DataFrame([])

for i in data.iterrows():
    if i['source'] in data_2['source'].values:
        joined_source.append(i)

joined_source = pd.DataFrame(joined_source)

joined_source.to_json("joined_source.json",orient="records",lines=True)
```
